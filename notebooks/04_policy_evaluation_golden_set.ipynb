# 04 - Policy Evaluation And Golden Set

Loads policy metrics and the deterministic Golden Set recommendations generated by the training runner.

```python
import json
from pathlib import Path

report_dir = Path('..') / 'reports' / 'policy_training'
metrics_path = report_dir / 'metrics.json'
golden_set_path = report_dir / 'golden_set_recommendations.json'

result = {}
if metrics_path.exists():
    result['metrics'] = json.loads(metrics_path.read_text(encoding='utf-8'))
if golden_set_path.exists():
    result['golden_set'] = json.loads(golden_set_path.read_text(encoding='utf-8'))
result or {'status': 'Run policy training first'}
```

# 02 - Data Processing and Validation

Transform the Hillstrom source file into ECloe's minimized offline-learning dataset and validate the generated schema.

## Pipeline Contract

- `segment` becomes the observed campaign `action`.
- `conversion` becomes the binary `reward`.
- Raw monetary history and ZIP code are excluded from the processed context.
- Validation writes a JSON report under `reports/data_validation.json`.

````python
from pathlib import Path

repo_root = Path.cwd()
if repo_root.name == 'notebooks':
    repo_root = repo_root.parent

raw_file = repo_root / 'data' / 'raw' / 'hillstrom.csv'
processed_file = repo_root / 'data' / 'processed' / 'hillstrom_processed.csv'
validation_report = repo_root / 'reports' / 'data_validation.json'

{
    'raw_file_exists': raw_file.exists(),
    'processed_file_exists': processed_file.exists(),
    'validation_report_exists': validation_report.exists(),
}
````
